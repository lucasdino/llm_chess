# Quick ipynb to convert latents_train to eval format

```python
import json
from pathlib import Path
```

```python
def reformat_latents(folder: str):
    latents_dir = Path(folder)
    for path in latents_dir.rglob('*.jsonl'):
        lines = path.read_text(encoding='utf-8').splitlines()
        new_lines = []
        for line in lines:
            data = json.loads(line)
            # original chat structure
            chat = data['chat']
            # extract user prompt and assistant response
            user_text      = chat[1][1]
            assistant_text = chat[2][1]
            # build new record
            new_record = {
                "chat": [
                    ["system", "chess_task_sysprompt.txt"],
                    ["user",   user_text],
                    ["assistant", ""]
                ],
                "info": {
                    "answer": {
                        "answer": assistant_text
                    }
                }
            }
            new_lines.append(json.dumps(new_record, ensure_ascii=False))
        # overwrite file with reformatted lines
        path.write_text('\n'.join(new_lines), encoding='utf-8')
```

```python
reformat_latents("latents_train")
```
